Neste momento temos uma nova demanda: criar um gráfico de linhas comparando os números de imigrantes do Brasil e Argentina para o Canadá, que são os maiores países da América do Sul.

Prepare-se para um mergulho fascinante nas linhas que conectam esses países da América do Sul ao território canadense. Nessa missão, a elaboração desse gráfico pode ser útil para a compreensão das tendências migratórias desses países para o Canadá ao longo do tempo e como elas se comparam entre si. Ao analisar esses fatores, podemos obter uma visão mais abrangente do cenário migratório na América do Sul.

Por isso, explore as diversas possibilidades e lembre-se dos elementos essenciais de um gráfico: título, rótulos nos eixos x e y e os ticks do eixo x, que devem ser definidos de 5 em 5 anos.

Além disso, você precisará descobrir como adicionar uma legenda para que seja possível identificar a linha de cada país. Ao seguir essas orientações, você terá construído um gráfico robusto que te permitirá uma análise significativa e aprofundada.

Após criar o gráfico analise o resultado obtido e reflita nas seguintes questões:

Há alguma tendência ou padrão comum nos dados dos dois países?
Quais são os períodos com maior número de imigrantes nos dois países?

```python
df_comparacao = df.loc[['Brasil', 'Argentina'], anos]
```

O DataFrame df_comparacao pode ser transposto para trocar as linhas pelas colunas e as colunas pelas linhas do DataFrame. Então, podemos usar a propriedade .T (abreviação para transpose):

```python
df_comparacao = df_comparacao.T

#Como resultado temos uma coluna com os anos e duas colunas para cada país contendo o número de imigrantes.

df_comparacao.head()
```

```python
plt.plot(df_comparacao['Brasil'],label = 'Brasil')
plt.plot(df_comparacao['Argentina'],label ='Argentina')
plt.title('Imigração do Brasil e Argentina para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.legend()
plt.show()
```

Com este gráfico, podemos notar que as tendências de imigração no Brasil e na Argentina eram bastante semelhantes no início do período. No entanto, a partir dos anos 2000, os números de imigrantes na Argentina começaram a aumentar e atingiram os valores mais elevados entre 2000 e 2005, antes de cair e atingir valores muito baixos nos anos mais recentes. Por outro lado, no caso do Brasil, os aumentos na imigração começaram após 2005 e parecem ter atingido o valor mais alto em 2010.

#### Comprando o processo de imigração para varios paises da america do sul

```python
fig, axs = plt.subplots(2,2, figsize=(10,6))
# Mudando o espaçamento vertical e horizontal
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Imigração dos quatro maiores países da America do Sul para o Canadá de 1980 a 2013')


axs[0,0].plot(df.loc['Brasil', anos])
axs[0,0].set_title('Brasil')

axs[0,1].plot(df.loc['Colômbia',anos])
axs[0,1].set_title('Colômbia')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Peru', anos])
axs[1,1].set_title('Peru')

#Melhorando a visualização do eixo x
for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))

# Adicionando titulos nos gráficos
for ax in axs.flat:
    ax.set_xlabel('Ano')
    ax.set_ylabel('Números de imigrantes')

ymin = 0
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)


plt.show()
```

### Adicionando a variavel ordenada

```python
# Plotar um gráfico que tenha o total de ligações pra cada um dos paises
# Utilizando um 'gráco de barras', ao invés de um gráfico de linhas, pois contém uma grande quantidade

#Lista de cores
cores = ['royalblue','orange','forestgreen', 'orchid','purple', 'brown','slateblue','gray','olive','navy','teal','tomato']

# Primeiro criamos uma figura e dizemos que ela contem eixos
# Essa figura tem um tamanho figsize
fig, ax = plt.subplots(figsize=(12,5))

# A linha abaixo determina quem é o meu x e y no gráfico, em seguida as cores
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)

#Setando um titulo, alinhamento e tamanho da fonte
ax.set_title('Imigração da America do Sul para o Canadá\n1980 a 2013', loc='left', fontsize=18)

ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()
```

### Destacando o Brasil no gráfico

```python
# Plotar um gráfico que tenha o total de ligações pra cada um dos paises
# Utilizando um 'gráco de barras na horizontal', ao invés de um gráfico de linhas, pois contém uma grande quantidade

#Lista de cores vazia
cores = []

# Fazendo uma interação para destacar o Brasil no gráfico
for pais in america_sul_sorted.index:
    if pais == 'Brasil':
        cores.append('green')
    else:
        cores.append('silver')

# Primeiro criamos uma figura e dizemos que ela contem eixos
# Essa figura tem um tamanho figsize
fig, ax = plt.subplots(figsize=(12,5))

# A linha abaixo determina quem é o meu x e y no gráfico, em seguida as cores
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)

#Setando um titulo, alinhamento e tamanho da fonte
ax.set_title('América do Sul: Brasil foi o quarto país com mais imigrantes\npara o Canadá no período de 1980 a 2013', loc='left', fontsize=18)

ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)



plt.show()
```

### Adicionando anotações nas barras do gráfico

```python
# Plotar um gráfico que tenha o total de ligações pra cada um dos paises
# Utilizando um 'gráco de barras na horizontal', ao invés de um gráfico de linhas, pois contém uma grande quantidade

#Lista de cores vazia
cores = []

# Fazendo uma interação para destacar o Brasil no gráfico
for pais in america_sul_sorted.index:
    if pais == 'Brasil':
        cores.append('green')
    else:
        cores.append('silver')

# Primeiro criamos uma figura e dizemos que ela contem eixos
# Essa figura tem um tamanho figsize
fig, ax = plt.subplots(figsize=(12,5))

# A linha abaixo determina quem é o meu x e y no gráfico, em seguida as cores
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)

#Setando um titulo, alinhamento e tamanho da fonte
ax.set_title('América do Sul: Brasil foi o quarto país com mais imigrantes\npara o Canadá no período de 1980 a 2013', loc='left', fontsize=18)

ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

# Adiconando mais uma iteração
for i, v in enumerate(america_sul_sorted['Total']):
    ax.text(v + 20, i, str(v), color='black', fontsize=10, ha='left', va='center')

# Retirando a caixa ao redor do gráfico
ax.set_frame_on(False)

# Retirando o eixo x
ax.get_xaxis().set_visible(False)

# Retirando algumas marcações no eixo Y ou eixo X
ax.tick_params(axis='both', which='both', length=0 )

# Salvando Gráfico
fig.savefig('imigracao_america_sul.png', transparent=False, dpi=300, bbox_inches='tight' )

plt.show()
```

### Adicionando uma paleta de cores

```python
def gerar_grafico_paleta(palette):
    fig, ax = plt.subplots(figsize=(8,4))
    ax = sns.barplot(data=top_10, y=top_10.index, x='Total', orient='h', palette=palette)
    ax.set_title('Países com maior imigração para o Canadá\n1980 a 2013', loc='left', fontsize=18)
    ax.set_xlabel('Número de imigrantes', fontsize = 14)
    ax.set_ylabel('')
    plt.show()
```

```python
#Paletas de cores sequenciais, muito usado para representar valores numericos
gerar_grafico_paleta('Blues')
```

```python
#Paletas de cores sequenciais, muito usado para representar valores numericos
#Com um argumento de deixar as cores mais escuras com paises com mior numero de imigrantes
gerar_grafico_paleta('Blues_r')
```

```python
gerar_grafico_paleta('rocket')
```

```python
# Essa paleta usada para representar dados que estão variando em torno de um valor central
gerar_grafico_paleta('coolwarm')
```

```python
# Essa paleta categorica, elas são usadas quando você tem diferentes categorias
gerar_grafico_paleta('tab10')
```

```python
# Usando outros temas
sns.set_theme(style='dark')
gerar_grafico_paleta('tab10')
```

```python
sns.set_theme(style='whitegrid')
gerar_grafico_paleta('tab10')
```
